```python
import pandas as pd
import numpy as np
import glob
import os
import re
import matplotlib.pyplot as plt

from hmmlearn.hmm import CategoricalHMM
from sklearn.model_selection import train_test_split
```

```python
def load_vote_file(file_path):
    df = pd.read_csv(
        file_path,
        skiprows=5,
        header=None,
        usecols=[0, 1, 2, 3],
        names=["Representative", "Party", "State", "Vote"]
    )

    for col in df.columns:
        df[col] = df[col].astype(str).str.strip()

    match = re.search(r"_(\d+)\.csv$", os.path.basename(file_path))
    df["Roll_Call"] = int(match.group(1)) if match else os.path.basename(file_path)

    vote_map = {
        "Yea": 1,
        "Nay": 0,
        "Present": 2,
        "Not Voting": 2
    }

    df["Vote_Code"] = df["Vote"].map(vote_map)
    df = df.dropna(subset=["Vote_Code"])
    df["Vote_Code"] = df["Vote_Code"].astype(int)

    return df
```

```python
base_folder = "data/raw" 
files = glob.glob(
    os.path.join(base_folder, "**", "*.csv"),
    recursive=True
)

print("Files found:", len(files))
print(files[:10])
```

```python
all_votes = pd.concat(
    [load_vote_file(file) for file in files],
    ignore_index=True
)

display(all_votes.head())
display(all_votes["Vote"].value_counts())
```

```python
vote_matrix = all_votes.pivot_table(
    index="Representative",
    columns="Roll_Call",
    values="Vote_Code",
    aggfunc="first"
)

rep_info = all_votes.groupby("Representative")[["Party", "State"]].first()

common_reps = vote_matrix.index.intersection(rep_info.index)

vote_matrix = vote_matrix.loc[common_reps]
rep_info = rep_info.loc[common_reps]

vote_matrix = vote_matrix.fillna(2).astype(int)

display(vote_matrix.head())
display(rep_info.head())

print("Vote matrix shape:", vote_matrix.shape)
```

```python
X = vote_matrix.to_numpy().reshape(-1, 1)
lengths = [vote_matrix.shape[1]] * vote_matrix.shape[0]

print("X shape:", X.shape)
print("Number of representatives:", vote_matrix.shape[0])
print("Votes per representative:", vote_matrix.shape[1])
print("Lengths sum:", sum(lengths))
```

EXPERIMENT 1: Hidden-state count test

```python
results = []

for k in [2, 3, 4, 5, 6]:
    model = CategoricalHMM(
        n_components=k,
        n_iter=200,
        random_state=42,
        tol=1e-4
    )

    model.fit(X, lengths)
    log_likelihood = model.score(X, lengths)

    results.append({
        "Hidden States": k,
        "Log Likelihood": log_likelihood,
        "AIC-like Score": -2 * log_likelihood + 2 * k
    })

results_df = pd.DataFrame(results)
display(results_df)

plt.figure(figsize=(7, 5))
plt.plot(results_df["Hidden States"], results_df["Log Likelihood"], marker="o")
plt.xlabel("Number of Hidden States")
plt.ylabel("Log Likelihood")
plt.title("Experiment 1: Model Fit vs Number of Hidden States")
plt.grid(True)
plt.show()
```

EXPERIMENT 2: Train/test prediction test

```python
from sklearn.model_selection import train_test_split

party_counts = rep_info.loc[vote_matrix.index, "Party"].value_counts()
display(party_counts)

parties = rep_info.loc[vote_matrix.index, "Party"]

if parties.value_counts().min() >= 2:
    train_reps, test_reps = train_test_split(
        vote_matrix.index,
        test_size=0.25,
        random_state=42,
        stratify=parties
    )
else:
    train_reps, test_reps = train_test_split(
        vote_matrix.index,
        test_size=0.25,
        random_state=42
    )

train_matrix = vote_matrix.loc[train_reps]
test_matrix = vote_matrix.loc[test_reps]

X_train = train_matrix.to_numpy().reshape(-1, 1)
lengths_train = [train_matrix.shape[1]] * train_matrix.shape[0]

X_test = test_matrix.to_numpy().reshape(-1, 1)
lengths_test = [test_matrix.shape[1]] * test_matrix.shape[0]

print("Train reps:", train_matrix.shape[0])
print("Test reps:", test_matrix.shape[0])
print("Train votes:", len(X_train))
print("Test votes:", len(X_test))

test_results = []

for k in [2, 3, 4, 5, 6]:
    model = CategoricalHMM(
        n_components=k,
        n_iter=200,
        random_state=42,
        tol=1e-4,
        init_params="ste"
    )

    model.fit(X_train, lengths_train)

    train_ll = model.score(X_train, lengths_train)
    test_ll = model.score(X_test, lengths_test)

    test_results.append({
        "Hidden States": k,
        "Train Log Likelihood": train_ll,
        "Test Log Likelihood": test_ll,
        "Train LL per Vote": train_ll / len(X_train),
        "Test LL per Vote": test_ll / len(X_test)
    })

test_results_df = pd.DataFrame(test_results)
display(test_results_df)

plt.figure(figsize=(7, 5))
plt.plot(
    test_results_df["Hidden States"],
    test_results_df["Train LL per Vote"],
    marker="o",
    label="Train"
)
plt.plot(
    test_results_df["Hidden States"],
    test_results_df["Test LL per Vote"],
    marker="o",
    label="Test"
)

plt.xlabel("Number of Hidden States")
plt.ylabel("Log Likelihood per Vote")
plt.title("Experiment 2: Train vs Test Performance")
plt.legend()
plt.grid(True)
plt.show()
```

EXPERIMENT 3: Transition matrix heatmap

```python
best_k = 3   # change this based on Experiment 1/2

best_model = CategoricalHMM(
    n_components=best_k,
    n_iter=200,
    random_state=42,
    tol=1e-4
)

best_model.fit(X, lengths)

transition_matrix = best_model.transmat_

plt.figure(figsize=(6, 5))
plt.imshow(transition_matrix)
plt.colorbar(label="Transition Probability")
plt.xticks(range(best_k), [f"State {i}" for i in range(best_k)])
plt.yticks(range(best_k), [f"State {i}" for i in range(best_k)])
plt.title("Experiment 3: Hidden State Transition Matrix")
plt.xlabel("Next Hidden State")
plt.ylabel("Current Hidden State")

for i in range(best_k):
    for j in range(best_k):
        plt.text(j, i, f"{transition_matrix[i, j]:.2f}", ha="center", va="center")

plt.show()

display(pd.DataFrame(
    transition_matrix,
    columns=[f"To State {i}" for i in range(best_k)],
    index=[f"From State {i}" for i in range(best_k)]
))
```

Expirement 4: Hidden state heatmap

```python
hidden_states = best_model.predict(X, lengths)

hidden_matrix = pd.DataFrame(
    hidden_states.reshape(vote_matrix.shape),
    index=vote_matrix.index,
    columns=vote_matrix.columns
)

display(hidden_matrix.head())

plt.figure(figsize=(12, 8))
plt.imshow(hidden_matrix, aspect="auto")
plt.colorbar(label="Hidden State")
plt.xlabel("Roll Call Vote")
plt.ylabel("Representative")
plt.title("Experiment 4: Inferred Hidden States Across Votes")
plt.show()
```

Experiment 5: Party vs hidden state

```python
# Most common hidden state for each representative
rep_hidden_state = hidden_matrix.mode(axis=1)[0]

party_state_df = pd.DataFrame({
    "Representative": rep_hidden_state.index,
    "Hidden State": rep_hidden_state.values,
    "Party": rep_info.loc[rep_hidden_state.index, "Party"].values
})

party_counts = pd.crosstab(
    party_state_df["Hidden State"],
    party_state_df["Party"]
)

display(party_counts)

party_counts.plot(kind="bar", figsize=(8, 5))
plt.xlabel("Hidden State")
plt.ylabel("Number of Representatives")
plt.title("Experiment 5: Party Distribution by Hidden State")
plt.xticks(rotation=0)
plt.show()
```
